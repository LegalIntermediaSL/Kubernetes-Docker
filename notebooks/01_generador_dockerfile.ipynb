# Generador basico de Dockerfiles

Este notebook sirve para construir plantillas simples de `Dockerfile` para proyectos Python. La utilidad es pedagogica: ayuda a ver que parametros cambian y cuales suelen mantenerse estables.


```python
from textwrap import dedent


def build_python_dockerfile(base_image="python:3.12-slim", workdir="/app", requirements_file="requirements.txt", app_file="app.py", port=8000, command=None):
    command = command or '["python", "app.py"]'
    return dedent(
        f"""
        FROM {base_image}

        WORKDIR {workdir}

        COPY {requirements_file} .
        RUN pip install --no-cache-dir -r {requirements_file}

        COPY {app_file} .

        EXPOSE {port}

        CMD {command}
        """
    ).strip()
```

```python
dockerfile = build_python_dockerfile(
    base_image="python:3.12-slim",
    app_file="app.py",
    port=8000,
)

print(dockerfile)
```
