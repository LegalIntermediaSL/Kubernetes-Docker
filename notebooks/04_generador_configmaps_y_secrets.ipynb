# Generador de ConfigMaps y Secrets

Este notebook ayuda a construir YAML simple para `ConfigMap` y `Secret`. La utilidad es doble: entender la estructura y generar prototipos antes de moverlos a `examples/`.


```python
from textwrap import dedent


def render_configmap(name, data):
    lines = [
        'apiVersion: v1',
        'kind: ConfigMap',
        'metadata:',
        f'  name: {name}',
        'data:',
    ]
    for key, value in data.items():
        lines.append(f'  {key}: "{value}"')
    return '\n'.join(lines)


def render_secret(name, string_data):
    lines = [
        'apiVersion: v1',
        'kind: Secret',
        'metadata:',
        f'  name: {name}',
        'type: Opaque',
        'stringData:',
    ]
    for key, value in string_data.items():
        lines.append(f'  {key}: "{value}"')
    return '\n'.join(lines)
```

```python
configmap_yaml = render_configmap(
    'app-config',
    {
        'APP_NAME': 'Curso Kubernetes',
        'APP_MODE': 'demo',
        'LOG_LEVEL': 'info',
    },
)

secret_yaml = render_secret(
    'app-secret',
    {
        'API_TOKEN': 'demo-token',
        'DB_PASSWORD': 'change-me',
    },
)

print(configmap_yaml)
print()
print(secret_yaml)
```

## Sugerencias de uso

- Genera primero la configuración aquí.
- Revisa nombres, claves y valores.
- Luego mueve el YAML final a `examples/k8s/configmap-secret/` o a un nuevo laboratorio.
