# Checklist de release y CI/CD

Este cuaderno genera una lista simple de verificacion para promocionar una version del curso.

```python
release = {
    "version": "1.2.0",
    "image_tag": "1.2.0",
    "target_environment": "demo",
    "checks": [
        "CI principal en verde",
        "docker-build en verde",
        "helm-validate en verde",
        "chart renderizado con values del entorno",
        "changelog actualizado",
        "plan de rollback identificado",
    ],
}

print(f"Release {release['version']} -> {release['target_environment']}")
for item in release["checks"]:
    print(f"- [ ] {item}")
print(f"\nTag de imagen esperada: {release['image_tag']}")
```
